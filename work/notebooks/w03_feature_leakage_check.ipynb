## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
## Build the Feature Vector
from sklearn.datasets import load_breast_cancer
import pandas as pd

# Load the Breast Cancer dataset
data = load_breast_cancer()

# Create DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Add target column
df["target"] = data.target

# Build feature vector and target
X = df.drop(columns=["target"])
y = df["target"]

# Fill missing values (good preprocessing practice)
X = X.fillna(X.median())

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)

X.head()
```

## Build the Feature Vector

I built the feature vector using the Breast Cancer dataset from scikit-learn.

The feature vector contains only predictor features, while the target column was separated before modeling.

The dataset contains only numeric features, so no categorical encoding was required.

Missing values were filled using the median as a standard preprocessing step, although this dataset does not contain missing values.

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
possible_leakage = []

keywords = ["target", "label", "result", "future", "outcome"]

for column in df.columns:
    column_name = column.lower()
    for keyword in keywords:
        if keyword in column_name:
            possible_leakage.append(column)

print("Possible leakage columns:", possible_leakage)
```

## Leakage Hunt

I checked the dataset for columns that could reveal the prediction target or contain future information.

The only detected column was the target column. It was removed before creating the feature vector.

No future information or label-derived features were observed.
