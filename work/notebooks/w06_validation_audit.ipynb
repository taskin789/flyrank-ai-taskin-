## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Leakage check
# Leakage check
# ML-09 Section 3 - Leakage Audit

possible_leakage = []

# Example feature names (replace these with your actual features if available)
feature_columns = [
    "avg_position",
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "page_length",
    "title_length"
]

keywords = [
    "target",
    "label",
    "future",
    "conversion",
    "outcome"
]

for col in feature_columns:
    if any(word in col.lower() for word in keywords):
        possible_leakage.append(col)

if possible_leakage:
    print("Potential leakage columns found:")
    print(possible_leakage)
else:
    print("No obvious leakage columns detected.")
```

## Leakage Audit

I checked the final feature set for possible target leakage by looking for columns that might contain target or future information.

The audit did not detect any obvious leakage columns based on the feature names.

This does not guarantee that leakage is impossible, but no clear evidence of leakage was observed in this check.
